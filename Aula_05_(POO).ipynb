# **Encapsulamento: Protegendo Dados e Comportamentos**

Esta etapa representa um aprofundamento no paradigma de Orientação a Objetos. O foco agora é em como proteger a integridade dos dados (atributos) e controlar o acesso aos comportamentos (métodos) de um objeto. O encapsulamento é um dos pilares da POO e promove a segurança e a robustez do código.

**Objetivo da Semana:** Compreender o conceito de encapsulamento e sua importância. Aprender a utilizar atributos privados para proteger o estado interno de um objeto e a implementar métodos públicos (como getters e setters) para fornecer acesso controlado. Ao final, o estudante será capaz de criar classes mais seguras, coesas e de fácil manutenção.

**Nota:** O encapsulamento não se trata de "esconder" a informação, mas sim de "controlar" o acesso a ela. Em vez de permitir que qualquer parte do código modifique os dados de um objeto livremente, criamos uma "interface" pública (um conjunto de métodos) que serve como ponto de entrada oficial para interagir com o objeto.

## **Mão na Massa**

Nesta etapa, vamos refatorar a classe `Musica` que desenvolvemos anteriormente. Atualmente, qualquer atributo de uma música, como `reproducoes`, por exemplo, pode ser alterado diretamente, o que pode levar a inconsistências. Por exemplo, veja a classe inicial:




````python
from datetime import datetime

class Musica:
  """
  Representa uma única música, encapsulando suas informações.
  """
  def __init__(self, titulo: str, artista: str, duracao: str, genero: str = "Desconhecido"):
    """
    Construtor da classe Musica.

    Args:
          titulo (str): O título da música.
          artista (str): O nome do artista ou banda.
          duracao (str): A duração da música no formato "M:SS".
          genero (str, optional): O gênero musical. Padrão "Desconhecido".
    """
    self.titulo = titulo
    self.artista = artista
    self.duracao = duracao
    self.genero = genero
    self.data_adicao = datetime.now()
    self.reproducoes = 0

  def reproduzir(self):
    """
    Simula a reprodução da música, incrementando o contador.
    """
    self.reproducoes += 1
    print(f"▶️  Reproduzindo: {self.titulo} de {self.artista}")
````

O exemplo a seguir demonstra a criação de uma Musica, a simulação de uma reprodução e o acesso ao atributo de reproduções.

````python
# Criando uma música
musica = Musica("Inesquecível", "Laura Pausini", "3:49", "Pop")
musica.reproduzir()
print(musica.reproducoes)  # Isso imprimirá 1
````

No entanto, o que impediria a adulteração do contador de reproduções?

````python
musica.reproducoes = -100  # Isso não deveria ser permitido!
print(musica.reproducoes)  # Resultado: -100 ❌
````

O **encapsulamento** resolve esse tipo de problema, pois nos permite **proteger atributos críticos da classe** e definir regras claras de acesso.

### **Atributos Privados**


Antes de refatorar a classe `Musica`, é necessário compreender como tratar atributos privados. Diferente de linguagens como Java ou C++, Python não disponibiliza modificadores de acesso explícitos, como `private` ou `public`. Em Python, utiliza-se uma convenção: atributos cujo identificador inicia com underscore (`_`) devem ser interpretados como de uso interno, ou seja, "privados" ou "protegidos".

**Nota:** O interpretador Python não impede o acesso, mas a comunidade de desenvolvedores entende e respeita essa convenção.

### **Classe Música com Encapsulamento**

A seguir, a classe `Musica` será reescrita em Python. O atributo `reproducoes` é renomeado para `_reproducoes`, e são definidos métodos específicos para sua leitura `get_reproducoes` e modificação `set_reproducoes` de forma controlada.

````python
from datetime import datetime

class Musica:
    """
    Representa uma única música, encapsulando suas informações de forma segura.
    """
    def __init__(self, titulo: str, artista: str, duracao: str, genero: str = "Desconhecido"):
        """
        Construtor da classe Musica.

        Args:
            titulo (str): O título da música.
            artista (str): O nome do artista ou banda.
            duracao (str): A duração da música no formato "M:SS".
            genero (str, optional): O gênero musical. Padrão "Desconhecido".
        """
        self.titulo = titulo
        self.artista = artista
        self.duracao = duracao
        self.genero = genero
        self.data_adicao = datetime.now()
        self._reproducoes = 0  # Atributo agora é "privado"

        print(f"Música '{self.titulo}' de {self.artista} foi criada com sucesso!")

    # Getter
    def get_reproducoes(self) -> int:
        """
        Retorna o número de vezes que a música foi reproduzida.
        """
        return self._reproducoes

    # Setter
    def set_reproducoes(self, valor: int):
        """
        Atualiza o número de reproduções da música de forma controlada.

        Args:
            valor (int): Novo valor para o contador de reproduções.

        Raises:
            ValueError: Se o valor informado for negativo.
        """
        if valor < 0:
            raise ValueError("O número de reproduções não pode ser negativo.")
        self._reproducoes = valor

    # Método que encapsula a lógica de reprodução
    def reproduzir(self):
        """
        Simula a reprodução da música, incrementando o contador de forma segura.
        """
        self._reproducoes += 1
        print(f"▶️  Reproduzindo: {self.titulo} de {self.artista} ({self._reproducoes}ª vez)")
````

**Observação:**

* O método `reproduzir()` constitui o único mecanismo oficial para incrementar o contador `_reproducoes`, centralizando a lógica e garantindo modificações previsíveis do estado interno.
* O método `get_reproducoes()` fornece acesso de leitura ao valor do atributo, sem permitir alterações externas diretas.
* O método `set_reproducoes(valor)` possibilita atualização controlada do contador, aplicando validações (por exemplo, impedindo valores negativos) e assegurando a consistência do objeto.

#### **Instanciando e Interagindo com Objetos Encapsulados**

Aqui está um exemplo de uso da classe `Musica` com os métodos `get_reproducoes`, `set_reproducoes` e `reproduzir`:

````python
# Criando uma música
musica = Musica("Imagine", "John Lennon", "3:15", genero="Rock")
````

````python
# Reproduzindo algumas vezes
musica.reproduzir()
musica.reproduzir()
````

````python
# Consultando o número de reproduções
print("Reproduções atuais:", musica.get_reproducoes())
````

````python
# Atualizando o número de reproduções manualmente
musica.set_reproducoes(0)
print("Reproduções atualizadas:", musica.get_reproducoes())
````

````python
# Tentando definir um valor inválido (negativo)
try:
    musica.set_reproducoes(-5)
except ValueError as e:
    print("Erro:", e)
````

**Atenção:** Mesmo com o uso de atributos privados (como _reproducoes) e métodos de acesso controlado, o interpretador Python não impede totalmente que o valor seja alterado externamente. O encapsulamento em Python funciona como um acordo de cavalheiros: a convenção _nome_do_atributo sinaliza aos desenvolvedores que o atributo é interno à classe e não deve ser acessado diretamente.

````python
# Embora seja possível, não é recomendado:
musica._reproducoes = -100
print(musica._reproducoes)  # Funciona, mas quebra a regra de integridade ❌
````

A efetividade do encapsulamento em Python depende do cumprimento da convenção pelos desenvolvedores. Ao aderir a essas práticas, assegura-se que os objetos sejam manipulados de forma consistente e previsível, preservando a integridade do estado interno e prevenindo comportamentos indesejados. Já em linguagens como Java ou C++, é possível declarar atributos como private, tornando o acesso externo impossível sem métodos específicos de acesso (getters/setters):

**Exemplo em Java:**

```java
public class Musica {
    private int reproducoes = 0; // atributo privado

    public int getReproducoes() {
        return reproducoes;
    }

    public void setReproducoes(int valor) {
        if (valor < 0) {
            throw new IllegalArgumentException("Reproduções não podem ser negativas");
        }
        reproducoes = valor;
    }
}

Musica m = new Musica();
m.reproducoes = -100; // ❌ Erro de compilação: reproducoes é privado
```

**Exemplo em C++:**

```java
class Musica {
private:
    int reproducoes = 0; // atributo privado

public:
    int getReproducoes() { return reproducoes; }
    void setReproducoes(int valor) {
        if(valor < 0) throw std::invalid_argument("Reproduções não podem ser negativas");
        reproducoes = valor;
    }
};

Musica m;
m.reproducoes = -100; // ❌ Erro: reproducoes é privado
```
