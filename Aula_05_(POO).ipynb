# **Encapsulamento**

Esta etapa representa um aprofundamento no paradigma de Orientação a Objetos. O foco agora é em como proteger a integridade dos dados (atributos) e controlar o acesso aos comportamentos (métodos) de um objeto. O encapsulamento é um dos pilares da POO e promove a segurança e a robustez do código.

**Objetivo da Semana:** Compreender o conceito de encapsulamento e sua importância. Aprender a utilizar atributos privados para proteger o estado interno de um objeto e a implementar métodos públicos (como getters e setters) para fornecer acesso controlado. Ao final, o estudante será capaz de criar classes mais seguras, coesas e de fácil manutenção.

**Nota:** O encapsulamento não se trata de "esconder" a informação, mas sim de "controlar" o acesso a ela. Em vez de permitir que qualquer parte do código modifique os dados de um objeto livremente, criamos uma "interface" pública (um conjunto de métodos) que serve como ponto de entrada oficial para interagir com o objeto.

## **Protegendo Dados e Comportamentos**

Nesta etapa, vamos refatorar a classe `Musica` que desenvolvemos anteriormente. Atualmente, qualquer atributo de uma música, como `reproducoes`, por exemplo, pode ser alterado diretamente, o que pode levar a inconsistências. Por exemplo, veja a classe inicial:




````python
from datetime import datetime

class Musica:
  """
  Representa uma única música, encapsulando suas informações.
  """
  def __init__(self, titulo: str, artista: str, duracao: str, genero: str = "Desconhecido"):
    """
    Construtor da classe Musica.

    Args:
          titulo (str): O título da música.
          artista (str): O nome do artista ou banda.
          duracao (str): A duração da música no formato "M:SS".
          genero (str, optional): O gênero musical. Padrão "Desconhecido".
    """
    self.titulo = titulo
    self.artista = artista
    self.duracao = duracao
    self.genero = genero
    self.data_adicao = datetime.now()
    self.reproducoes = 0

  def reproduzir(self):
    """
    Simula a reprodução da música, incrementando o contador.
    """
    self.reproducoes += 1
    print(f"▶️  Reproduzindo: {self.titulo} de {self.artista}")
````

O exemplo a seguir demonstra a criação de uma Musica, a simulação de uma reprodução e o acesso ao atributo de reproduções.

````python
# Criando uma música
musica = Musica("Inesquecível", "Laura Pausini", "3:49", "Pop")
musica.reproduzir()
print(musica.reproducoes)  # Isso imprimirá 1
````

No entanto, o que impediria a adulteração do contador de reproduções?

````python
musica.reproducoes = -100  # Isso não deveria ser permitido!
print(musica.reproducoes)  # Resultado: -100 ❌
````

O **encapsulamento** resolve esse tipo de problema, pois nos permite **proteger atributos críticos da classe** e definir regras claras de acesso.

### **Atributos Privados**


Antes de refatorar a classe `Musica`, é necessário compreender como tratar atributos privados. Diferente de linguagens como Java ou C++, Python não disponibiliza modificadores de acesso explícitos, como `private` ou `public`. Em Python, utiliza-se uma convenção: atributos cujo identificador inicia com underscore (`_`) devem ser interpretados como de uso interno, ou seja, "privados" ou "protegidos".

**Nota:** O interpretador Python não impede o acesso, mas a comunidade de desenvolvedores entende e respeita essa convenção.

### **Classe Música com Encapsulamento**

A seguir, a classe `Musica` será reescrita em Python. O atributo `reproducoes` é renomeado para `_reproducoes`, e são definidos métodos específicos para sua leitura `get_reproducoes` e modificação `set_reproducoes` de forma controlada.

````python
from datetime import datetime

class Musica:
    """
    Representa uma única música, encapsulando suas informações de forma segura.
    """
    def __init__(self, titulo: str, artista: str, duracao: str, genero: str = "Desconhecido"):
        """
        Construtor da classe Musica.

        Args:
            titulo (str): O título da música.
            artista (str): O nome do artista ou banda.
            duracao (str): A duração da música no formato "M:SS".
            genero (str, optional): O gênero musical. Padrão "Desconhecido".
        """
        self.titulo = titulo
        self.artista = artista
        self.duracao = duracao
        self.genero = genero
        self.data_adicao = datetime.now()
        self._reproducoes = 0  # Atributo agora é "privado"

        print(f"Música '{self.titulo}' de {self.artista} foi criada com sucesso!")

    # Getter
    def get_reproducoes(self) -> int:
        """
        Retorna o número de vezes que a música foi reproduzida.
        """
        return self._reproducoes

    # Setter
    def set_reproducoes(self, valor: int):
        """
        Atualiza o número de reproduções da música de forma controlada.

        Args:
            valor (int): Novo valor para o contador de reproduções.

        Raises:
            ValueError: Se o valor informado for negativo.
        """
        if valor < 0:
            raise ValueError("O número de reproduções não pode ser negativo.")
        self._reproducoes = valor

    # Método que encapsula a lógica de reprodução
    def reproduzir(self):
        """
        Simula a reprodução da música, incrementando o contador de forma segura.
        """
        self._reproducoes += 1
        print(f"▶️  Reproduzindo: {self.titulo} de {self.artista} ({self._reproducoes}ª vez)")
````

**Observação:**

* O método `reproduzir()` constitui o único mecanismo oficial para incrementar o contador `_reproducoes`, centralizando a lógica e garantindo modificações previsíveis do estado interno.
* O método `get_reproducoes()` fornece acesso de leitura ao valor do atributo, sem permitir alterações externas diretas.
* O método `set_reproducoes(valor)` possibilita atualização controlada do contador, aplicando validações (por exemplo, impedindo valores negativos) e assegurando a consistência do objeto.

### **Properties: A Forma Pythônica de Gerenciar Atributos**


A abordagem com métodos `get_` e `set_` é eficaz e comum em outras linguagens, como Java. No entanto, em Python, existe uma maneira de obter o mesmo nível de controle e segurança, mas com uma sintaxe muito mais limpa e intuitiva. O objetivo é permitir que o acesso aos atributos se pareça com um acesso direto, mas que, por baixo dos panos, métodos de validação (getters e setters) sejam executados.

#### **O Decorador @property (Getter)**

O decorador `@property` permite expor um método como um atributo gerenciado de leitura apenas, funcionando como um getter. A seguir, a classe `Musica` será refatorada para converter o método `get_reproducoes` em uma `property`.

> **Nota:** Um decorador em Python é uma função que recebe outra função ou método como argumento e retorna uma nova função ou método, podendo modificar ou estender seu comportamento sem alterar seu código original.

````python
# ... (início da classe Musica) ...

@property
def reproducoes(self) -> int:
  """
  Getter para o atributo _reproducoes.
  Agora é acessado como 'musica.reproducoes'.
  """
  return self._reproducoes

# ... (restante da classe) ...
````

Com essa mudança, o método reproducoes pode ser acessado como um atributo, sem os parênteses: `musica.reproducoes`.

#### **O Decorador `@<nome_da_property>.setter` (Setter)**



Para complementar a property e permitir a alteração do valor (criando um setter), usamos um decorador que deriva do nome da property que acabamos de criar. Para a property reproducoes, o decorador do setter será `@reproducoes.setter`.

````python
# ... (início da classe e a @property) ...

@reproducoes.setter
def reproducoes(self, valor: int):
  """
  Setter para o atributo _reproducoes.
  Valida o valor antes de atribuí-lo.
  """
  if valor < 0:
    raise ValueError("O número de reproduções não pode ser negativo.")
  print(f"(Setter) O valor de _reproducoes foi alterado para {valor}.")
  self._reproducoes = valor

# ... (restante da classe) ...
````

A atribuição de um valor ao atributo `musica.reproducoes` aciona automaticamente o método `reproducoes` decorado com `@reproducoes.setter`. Esse método gerencia o acesso à variável interna, aplicando a lógica de validação definida, garantindo que apenas valores válidos sejam armazenados e mantendo a integridade do estado do objeto.



#### **Classe Musica Refatorada com @property**

A seguir, apresenta-se a implementação da classe utilizando `properties`. Observe que `get_reproducoes` e `set_reproducoes` foram substituídos pelo uso de `@property` e `@<property>.setter`, promovendo encapsulamento e controle de acesso aos atributos.


````python
from datetime import datetime

class Musica:
    """
    Representa uma única música, utilizando properties para um encapsulamento Pythônico.
    """
    def __init__(self, titulo: str, artista: str, duracao: str, genero: str = "Desconhecido"):
        self.titulo = titulo
        self.artista = artista
        self.duracao = duracao
        self.genero = genero
        self.data_adicao = datetime.now()
        self._reproducoes = 0

        print(f"Música '{self.titulo}' de {self.artista} foi criada com sucesso!")

    @property
    def reproducoes(self) -> int:
        """Getter para o atributo _reproducoes."""
        return self._reproducoes

    @reproducoes.setter
    def reproducoes(self, valor: int):
        """Setter para o atributo _reproducoes com validação."""
        if not isinstance(valor, int) or valor < 0:
            raise ValueError("O número de reproduções deve ser um inteiro não-negativo.")
        self._reproducoes = valor

    def reproduzir(self):
        """
        Simula a reprodução da música, incrementando o contador de forma segura.
        """
        self._reproducoes += 1 # Acesso interno ao atributo privado
        print(f"▶️  Reproduzindo: {self.titulo} de {self.artista} ({self._reproducoes}ª vez)")
````

A interação com a classe agora é muito mais natural e legível, mantendo toda a segurança que havíamos implementado.

````python
# Criando uma música
musica_prop = Musica("Bohemian Rhapsody", "Queen", "5:55", genero="Rock")
````

````python
# Reproduzindo algumas vezes
musica_prop.reproduzir()
musica_prop.reproduzir()
````

````python
# Consultando o número de reproduções (chama o @property getter)
print(f"Reproduções atuais: {musica_prop.reproducoes}")
````

````python
# Atualizando o número de reproduções (chama o @reproducoes.setter)
print("\nResetando as reproduções...")
musica_prop.reproducoes = 0
print(f"Reproduções atualizadas: {musica_prop.reproducoes}")
````

````python
# Tentando definir um valor inválido (negativo)
try:
    print("\nTentando atribuir um valor inválido...")
    musica_prop.reproducoes = -10
except ValueError as e:
    print(f"Erro ao tentar atualizar: {e}")
print(f"Reproduções finais: {musica_prop.reproducoes}")
````

**Vantagens de Utilizar @property**

- **API Limpa e Intuitiva:** Permite acessar e modificar atributos de forma natural (objeto.atributo = valor), evitando chamadas explícitas a métodos como objeto.set_atributo(valor). Isso torna o código mais legível e fácil de entender.

- **Encapsulamento Preservado:** A lógica de validação, restrição ou cálculo continua protegida, mas é exposta de forma transparente através de uma interface de atributo simples.

- **Flexibilidade para Evolução do Código:** É possível começar com atributos públicos simples e, posteriormente, adicionar lógica de validação ou processamento ao transformá-los em properties, sem alterar a forma como o código cliente interage com o objeto.
