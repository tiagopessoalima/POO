# **Classes Abstratas**

As **Classes Abstratas** são classes que não podem ser instanciadas diretamente. Elas servem como **modelos** ou **contratos** para outras classes. O principal objetivo de uma classe abstrata é definir uma interface comum que suas subclasses devem seguir, garantindo que todas as classes derivadas implementem um conjunto específico de métodos.

**Objetivo da Semana:** Compreender a necessidade de classes abstratas para definir interfaces obrigatórias e como implementá-las em Python usando o módulo `abc`. Desenvolver a capacidade de criar contratos de código que garantam que as classes derivadas implementem os métodos necessários.

**Nota:** Em Python, o conceito de classes abstratas é implementado através do módulo `abc` (Abstract Base Classes).

## **Por que usar Classes Abstratas?**


### **O Problema: Falta de Padronização**


Suponha que temos diferentes tipos de repositórios sem um contrato comum:

````python
class RepositorioMemoria:
    def salvar(self, item):
        # Implementação específica
        pass
````

````python
class RepositorioArquivo:
    def guardar(self, item):  # Nome diferente!
        # Implementação diferente
        pass
````

Problemas:

- Métodos com nomes diferentes para a mesma funcionalidade

- Dificuldade em trocar implementações

- Código cliente precisa conhecer detalhes específicos de cada implementação

## **A Solução: Contratos com ABCs**


````python
from abc import ABC, abstractmethod

class Repositorio(ABC):
    @abstractmethod
    def salvar(self, item): pass

    @abstractmethod
    def buscar_por_id(self, id): pass

    @abstractmethod
    def listar_todos(self): pass
````

Agora todas as implementações devem seguir o mesmo contrato!

## **Implementando Classes Abstratas em Python**

Para transformar uma classe em uma Classe Abstrata em Python, precisamos seguir dois passos:

1.  Herdar da classe `ABC` (Abstract Base Class) do módulo `abc`.
2.  Decorar os métodos que devem ser obrigatórios nas subclasses com `@abstractmethod`.

### **Exemplo Prático: Repositório de Dados**

Vamos criar uma classe abstrata `Repositorio` que define o contrato básico para qualquer classe que lide com persistência de dados (salvar, buscar, deletar), independentemente de ser um banco de dados, um arquivo ou um serviço web.

````python
from abc import ABC, abstractmethod
from typing import Any, List, Dict, Optional

class Repositorio(ABC):
    """
    Classe Abstrata Base para todos os Repositórios de Dados.
    Define um contrato (interface) que todas as subclasses devem seguir.

    Princípios SOLID aplicados:
    - S: Single Responsibility - Define apenas o contrato para repositórios
    - O: Open/Closed - Aberta para extensão, fechada para modificação
    - L: Liskov Substitution - Qualquer subclasse pode substituir a base
    - I: Interface Segregation - Interface coesa e focada
    - D: Dependency Inversion - Dependa de abstrações, não de implementações
    """

    @abstractmethod
    def salvar(self, item: Any) -> int:
        """
        Método abstrato para salvar um item.
        Retorna o ID do item salvo.
        Obrigatório ser implementado nas subclasses.
        """
        pass

    @abstractmethod
    def buscar_por_id(self, id: int) -> Optional[Any]:
        """
        Método abstrato para buscar um item pelo ID.
        Retorna o item ou None se não encontrado.
        Obrigatório ser implementado nas subclasses.
        """
        pass

    @abstractmethod
    def listar_todos(self) -> List[Any]:
        """
        Método abstrato para listar todos os itens.
        Obrigatório ser implementado nas subclasses.
        """
        pass

    @abstractmethod
    def deletar(self, id: int) -> bool:
        """
        Método abstrato para deletar um item pelo ID.
        Retorna True se foi deletado, False caso contrário.
        Obrigatório ser implementado nas subclasses.
        """
        pass

    def conectar(self) -> None:
        """
        Método concreto (não abstrato) que pode ter uma implementação padrão
        ou ser usado como um utilitário.
        Hook method: subclasses podem sobrescrever se necessário.
        """
        print("🔗 Conectando ao recurso de dados...")

    def desconectar(self) -> None:
        """
        Outro método concreto com implementação padrão.
        """
        print("🔌 Desconectando do recurso de dados...")
````

### **1. Tentativa de Instanciação**

Se tentarmos instanciar a classe abstrata `Repositorio` diretamente, Python nos impedirá, pois ela é apenas um modelo.

````python
# Tentar instanciar a classe abstrata
try:
    repo_base = Repositorio()
except TypeError as e:
    print(f"❌ Erro ao instanciar Repositorio: {e}")
````

### **2. Implementando o Contrato: Repositório em Memória**

Agora, vamos criar uma subclasse concreta que herda de `Repositorio` e implementa **todos** os métodos abstratos.

````python
class RepositorioEmMemoria(Repositorio):
    """
    Implementação concreta do Repositorio usando um dicionário em memória.
    Útil para testes e protótipos.
    """

    def __init__(self):
        self._dados: Dict[int, Any] = {}
        self._proximo_id = 1
        self.conectar()  # Chama o método concreto da classe base

    def salvar(self, item: Any) -> int:
        """
        Implementação concreta de salvar: adiciona ao dicionário.
        """
        item_id = self._proximo_id
        item_com_id = {"id": item_id, "valor": item, "timestamp": "2024-01-01"}
        self._dados[item_id] = item_com_id
        print(f"✅ Item salvo com ID: {item_id}")
        self._proximo_id += 1
        return item_id

    def buscar_por_id(self, id: int) -> Optional[Any]:
        """
        Implementação concreta de buscar_por_id.
        """
        return self._dados.get(id, None)

    def listar_todos(self) -> List[Any]:
        """
        Implementação concreta de listar_todos.
        """
        return list(self._dados.values())

    def deletar(self, id: int) -> bool:
        """
        Implementação concreta de deletar.
        """
        if id in self._dados:
            del self._dados[id]
            print(f"🗑️ Item com ID {id} deletado")
            return True
        print(f"⚠️ Item com ID {id} não encontrado")
        return False

    # Podemos sobrescrever métodos concretos se necessário
    def conectar(self) -> None:
        print("🧠 Inicializando repositório em memória...")

# Testando a implementação concreta
print("=== TESTE REPOSITÓRIO EM MEMÓRIA ===")
repo_memoria = RepositorioEmMemoria()

# Testando operações
id1 = repo_memoria.salvar("Primeiro Objeto")
id2 = repo_memoria.salvar("Segundo Objeto")

print("\n📋 Listando todos:")
for item in repo_memoria.listar_todos():
    print(f"  - {item}")

print(f"\n🔍 Buscando ID {id1}:")
print(f"  {repo_memoria.buscar_por_id(id1)}")

print(f"\n🔍 Buscando ID 999 (inexistente):")
print(f"  {repo_memoria.buscar_por_id(999)}")

print(f"\n🗑️ Deletando ID {id1}:")
repo_memoria.deletar(id1)

print(f"\n📋 Listando após deleção:")
for item in repo_memoria.listar_todos():
    print(f"  - {item}")
````

### **3. Outra Implementação: Repositório em Arquivo**

Vamos criar outra implementação para demonstrar o poder das ABCs:

````python
import json
import os

class RepositorioEmArquivo(Repositorio):
    """
    Implementação concreta do Repositorio usando arquivo JSON.
    """

    def __init__(self, arquivo: str = "dados.json"):
        self.arquivo = arquivo
        self._dados = {}
        self._proximo_id = 1
        self._carregar_dados()
        self.conectar()

    def _carregar_dados(self):
        """Método privado para carregar dados do arquivo."""
        if os.path.exists(self.arquivo):
            with open(self.arquivo, 'r', encoding='utf-8') as f:
                dados_carregados = json.load(f)
                self._dados = {int(k): v for k, v in dados_carregados.items()}
                if self._dados:
                    self._proximo_id = max(self._dados.keys()) + 1

    def _salvar_dados(self):
        """Método privado para salvar dados no arquivo."""
        with open(self.arquivo, 'w', encoding='utf-8') as f:
            json.dump(self._dados, f, indent=2, ensure_ascii=False)

    def salvar(self, item: Any) -> int:
        item_id = self._proximo_id
        self._dados[item_id] = {
            "id": item_id,
            "valor": item,
            "timestamp": "2024-01-01"
        }
        self._proximo_id += 1
        self._salvar_dados()
        print(f"💾 Item salvo no arquivo com ID: {item_id}")
        return item_id

    def buscar_por_id(self, id: int) -> Optional[Any]:
        return self._dados.get(id, None)

    def listar_todos(self) -> List[Any]:
        return list(self._dados.values())

    def deletar(self, id: int) -> bool:
        if id in self._dados:
            del self._dados[id]
            self._salvar_dados()
            print(f"🗑️ Item com ID {id} deletado do arquivo")
            return True
        return False

    def conectar(self) -> None:
        print(f"📁 Conectando ao arquivo: {self.arquivo}")

# Testando a implementação em arquivo
print("\n=== TESTE REPOSITÓRIO EM ARQUIVO ===")
repo_arquivo = RepositorioEmArquivo("teste.json")

id_arquivo = repo_arquivo.salvar("Dados persistentes")
encontrado = repo_arquivo.buscar_por_id(id_arquivo)
print(f"Item encontrado: {encontrado}")

repo_arquivo.deletar(id_arquivo)
````

### **4. Falha ao Implementar o Contrato**

Se uma subclasse herdar de uma ABC, mas falhar em implementar todos os métodos abstratos, Python também gerará um erro. Isso é o que garante o contrato.

````python
class RepositorioIncompleto(Repositorio):
    """
    Esta classe falha em implementar o método 'listar_todos' e 'deletar'.
    """
    def salvar(self, item: Any) -> int:
        print("Salvando...")
        return 1

    def buscar_por_id(self, id: int) -> Any:
        return None

    # Os métodos 'listar_todos' e 'deletar' estão faltando!

# Tentar instanciar a classe incompleta
try:
    repo_incompleto = RepositorioIncompleto()
except TypeError as e:
    print(f"\n❌ Erro ao instanciar RepositorioIncompleto: {e}")
````

## **Vantagens e Casos de Uso**

Classes Abstratas são ferramentas poderosas para design de software robusto e manutenível.

| Vantagem | Descrição |
| :--- | :--- |
| **Garantia de Interface** | Força as classes derivadas a terem uma assinatura de método específica, evitando que desenvolvedores esqueçam de implementar funcionalidades cruciais. |
| **Padrões de Projeto** | São a base para a implementação de diversos padrões de projeto, como o *Template Method* e o *Factory Method*. |
| **Consistência** | Permite que o código cliente trate objetos de diferentes classes derivadas de forma uniforme, sabendo que o método obrigatório estará sempre presente. |
| **Organização** | Deixa claro qual é o papel da classe base (definir a interface) e qual é o papel das subclasses (fornecer a implementação). |

### **Caso de Uso: Frameworks e Plugins**

Em frameworks ou sistemas de plugins, as Classes Abstratas são essenciais. Por exemplo, um framework de processamento de arquivos pode definir uma `ProcessadorDeArquivo(ABC)` com um método abstrato `processar()`. Qualquer plugin que queira se integrar ao framework deve, obrigatoriamente, criar uma subclasse e implementar o método `processar()`.

````python
# Exemplo de Framework
class Plugin(ABC):
    @abstractmethod
    def executar(self, dados: dict) -> dict:
        """Contrato: todo plugin deve ter um método executar."""
        pass

# Plugin de Criptografia
class PluginCriptografia(Plugin):
    def executar(self, dados: dict) -> dict:
        print("🔑 Criptografando dados...")
        # Lógica de criptografia
        return {"resultado": "dados_criptografados"}

# Plugin de Log
class PluginLog(Plugin):
    def executar(self, dados: dict) -> dict:
        print("📝 Registrando log...")
        # Lógica de log
        return dados

# O código cliente usa os plugins de forma polimórfica, garantindo que 'executar' existe
plugins = [PluginCriptografia(), PluginLog()]
dados_iniciais = {"user": "Alice"}

for plugin in plugins:
    dados_iniciais = plugin.executar(dados_iniciais)

print(f"\nDados finais após processamento: {dados_iniciais}")
````

## **Exercício Prático**

Crie uma hierarquia de classes para diferentes tipos de animais.

1.  Crie uma Classe Abstrata chamada `Animal` que herde de `ABC`.
2.  Defina um método abstrato chamado `emitir_som(self)`.
3.  Defina um método concreto chamado `mover(self)` que imprima "O animal está se movendo.".
4.  Crie as classes concretas `Cachorro` e `Gato` que herdem de `Animal`.
5.  Implemente o método `emitir_som` em cada subclasse para imprimir o som correto.
6.  Tente instanciar a classe `Animal` e observe o erro.
7.  Instancie `Cachorro` e `Gato` e chame os métodos `emitir_som` e `mover`.

````python
# Seu código de exercício aqui
from abc import ABC, abstractmethod

# 1. Crie a Classe Abstrata Animal
class Animal(ABC):

    # 3. Defina um método concreto
    def mover(self):
        print("O animal está se movendo.")

    # 2. Defina o método abstrato
    @abstractmethod
    def emitir_som(self):
        pass

# 4. Crie as classes concretas Cachorro e Gato
class Cachorro(Animal):
    # 5. Implemente o método abstrato
    def emitir_som(self):
        print("Au Au!")

class Gato(Animal):
    # 5. Implemente o método abstrato
    def emitir_som(self):
        print("Miau!")

# 6. Tente instanciar a classe Animal
try:
    animal_generico = Animal()
except TypeError as e:
    print(f"\nErro (esperado) ao instanciar Animal: {e}")

# 7. Instancie e teste as classes concretas
print("\n--- Testando Cachorro ---")
cachorro = Cachorro()
cachorro.mover()
cachorro.emitir_som()

print("\n--- Testando Gato ---")
gato = Gato()
gato.mover()
gato.emitir_som()
````
