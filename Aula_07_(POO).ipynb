# **Herança**

Esta etapa aprofunda o estudo do paradigma de POO, com ênfase na construção de **hierarquias de classes** que permitem a reutilização de código e a modelagem de relacionamentos do tipo *“é um”* (*is-a*). Nesse contexto, uma **subclasse** (ou classe derivada) pode herdar atributos e métodos de uma **superclasse** (ou classe base), possibilitando a especialização ou a extensão de comportamentos previamente definidos. A **herança**, enquanto um dos pilares fundamentais da POO, favorece a reutilização sistemática de código e a organização estrutural de sistemas orientados a objetos.

**Objetivo da Semana:** Assimilar o conceito de **herança** e compreender sua relevância para a **reutilização e organização de código**. Desenvolver a capacidade de implementar **subclasses** (classes derivadas) que herdam atributos e métodos de uma **superclasse** (classe base), aplicar a **sobrescrita de métodos** (*method overriding*) para especialização de comportamento e utilizar a função `super()` para estender funcionalidades já existentes. Ao término, o estudante deverá ser capaz de projetar sistemas orientados a objetos com **menor redundância** e **maior coesão estrutural**.

**Nota:** A herança define um relacionamento do tipo *“é um”* (*is-a*). Por exemplo, a classe `Carro` pode ser modelada como uma especialização da classe `Veiculo`, e a classe `Gato` como uma especialização de `Animal`. Esse tipo de relação possibilita que instâncias de subclasses sejam tratadas como instâncias de suas superclasses, o que estabelece a base para o conceito de **polimorfismo**.

## **Reutilizando Código com Classes**

Vamos iniciar com a definição da classe **`Funcionario`**, utilizada em um sistema de Recursos Humanos. Essa classe representa um funcionário genérico da organização, servindo como **superclasse** para futuras especializações.

````python
class Funcionario:
  """
  Representa um funcionário genérico da empresa.
  """
  def __init__(self, nome: str, cpf: str, salario: float):
    """
    Construtor da classe Funcionario.

    Args:
        nome (str): O nome completo do funcionário.
        cpf (str): O CPF do funcionário.
        salario (float): O salário base do funcionário.
    """
    self.nome = nome
    self.cpf = cpf
    self.salario = salario

  def exibir_dados(self):
    """
    Exibe os dados básicos do funcionário.
    """
    print(f"Nome: {self.nome}")
    print(f"CPF: {self.cpf}")
    print(f"Salário: R$ {self.salario:.2f}")

  def calcular_bonus(self) -> float:
    """
    Calcula o bônus padrão para um funcionário (10% do salário).
    """
    return self.salario * 0.10
````

Agora, vamos testar a nossa classe `Funcionario`:

````python
# Criando um funcionário
func_comum = Funcionario("Carlos Silva", "123.456.789-00", 3000.00)
func_comum.exibir_dados()
print(f"Bônus a receber: R$ {func_comum.calcular_bonus():.2f}")
````

Suponha que seja necessário modelar um objeto `Gerente`, que também representa um funcionário, mas com responsabilidades adicionais, como autenticação via senha de acesso ao sistema e um esquema de bonificação diferenciado (10% do salário + R\$500). **Sem recorrer à herança**, seria preciso **duplicar grande parte da implementação já existente na classe `Funcionario`**:

````python
class Gerente:
    """
    Representa um gerente da empresa.
    """
    def __init__(self, nome: str, cpf: str, salario: float, senha: str):
        """
        Construtor da classe Gerente.

        Args:
            nome (str): O nome completo do gerente.
            cpf (str): O CPF do gerente.
            salario (float): O salário base do gerente.
            senha (str): Senha de acesso ao sistema.
        """
        self.nome = nome
        self.cpf = cpf
        self.salario = salario
        self.senha = senha

    def exibir_dados(self):
        """
        Exibe os dados básicos do gerente.
        """
        print(f"Nome: {self.nome}")
        print(f"CPF: {self.cpf}")
        print(f"Salário: R$ {self.salario:.2f}")

    def calcular_bonus(self) -> float:
        """
        Calcula o bônus para um gerente (15% do salário + R$ 500).
        """
        return self.salario * 0.15 + 500

    def autenticar(self, senha_tentativa: str) -> bool:
        """
        Verifica se a senha fornecida é a correta.
        """
        if self.senha == senha_tentativa:
            print("Acesso concedido!")
            return True
        else:
            print("Acesso negado.")
            return False
````

Agora, vamos testar a classe `Gerente`

````python
# Criando um gerente
gerente = Gerente("Ana Paula", "987.654.321-00", 8000.00, "senha123")
gerente.exibir_dados()
print(f"Bônus a receber: R$ {gerente.calcular_bonus():.2f}")
````

> **Nota:** Perceba que a classe `Gerente` repete quase todo o código da classe `Funcionario`. Se fosse necessário alterar a forma de exibir os dados, por exemplo, teríamos que modificar em duas classes diferentes, aumentando a chance de erros e inconsistências. É exatamente aqui que entra a **herança**, permitindo **reaproveitar código** sem duplicação.

### **Criando uma Classe Filha (Subclasse)**


Em **Python**, a herança é declarada colocando o nome da superclasse entre parênteses na definição da **subclasse**:

```python
class Gerente(Funcionario):
    # corpo da classe
    pass
```

Nesse caso, Gerente herda todos os atributos e métodos de `Funcionario`, podendo sobrescrevê-los ou estendê-los. Já em **Java**, a herança é estabelecida por meio da palavra-chave `extends`:

```java
public class Gerente extends Funcionario {
    // corpo da classe
}
```

Assim como em Python, a classe `Gerente` em Java herda todos os membros (atributos e métodos) definidos em `Funcionario`. A principal distinção está no modelo de herança adotado por cada linguagem: em **Java**, a herança é **estritamente simples**, de modo que uma classe só pode estender diretamente uma única superclasse. Já em **Python**, o mecanismo permite **herança múltipla**, possibilitando que uma subclasse derive de mais de uma classe base simultaneamente. Essa diferença impacta diretamente no desenho da hierarquia de classes e nas estratégias de reutilização de código em cada linguagem.


### **Sobrescrevendo e Estendendo Funcionalidades**

A seguir, a classe `Gerente` será criada como uma **subclasse** de `Funcionario`.
Ela herda todos os membros da classe base, **sobrescreve** métodos para especializar comportamentos e **estende** a funcionalidade ao incluir novos atributos e operações específicas.

````python
# Classe Filha
class Gerente(Funcionario): # Gerente herda de Funcionario
    """
    Representa um Gerente, que é um tipo de Funcionário com mais responsabilidades.
    """
    def __init__(self, nome: str, cpf: str, salario: float, senha: str):
        """
        Construtor da classe Gerente.

        Args:
            nome (str): O nome completo do gerente.
            cpf (str): O CPF do gerente.
            salario (float): O salário base do gerente.
            senha (str): A senha do gerente para acesso ao sistema.
        """
        # Chama o construtor da classe mãe (Funcionario) para inicializar
        # os atributos que já existem lá.
        super().__init__(nome, cpf, salario)
        self.senha = senha

    # Redefinindo (Sobrescrevendo) o método da classe mãe
    def calcular_bonus(self) -> float:
        """
        Calcula o bônus para um gerente (15% do salário + R$ 500).
        """
        bonus_base = super().calcular_bonus() # Podemos usar o cálculo da mãe
        return self.salario * 0.15 + 500

    def autenticar(self, senha_tentativa: str) -> bool:
        """
        Verifica se a senha fornecida é a correta.
        """
        if self.senha == senha_tentativa:
            print("Acesso concedido!")
            return True
        else:
            print("Acesso negado.")
            return False
````

**Observação:**

- A sintaxe `class Gerente(Funcionario):` estabelece a herança.

- A função `super().__init__(...)` é usada para chamar o construtor da classe mãe (`Funcionario`) e evitar a reescrita da lógica de inicialização de `nome`, `cpf` e `salario`.

- O método `calcular_bonus` foi **sobrescrito**. A classe Gerente tem sua própria versão desse método, pois a regra de negócio é diferente.

- Adicionamos um método novo, `autenticar`, que só existe na classe `Gerente`.

### **Herança Múltipla: Combinando Funcionalidades**


Enquanto a herança simples modela uma relação "é um" com uma única classe mãe (um `Gerente` é um `Funcionario`), a **herança múltipla** permite que uma classe filha herde atributos e métodos de **duas ou mais classes** mãe. Isso é útil para combinar funcionalidades de diferentes fontes, criando uma classe que é uma fusão de várias outras. A sintaxe é uma extensão natural da herança simples: basta listar todas as classes mãe entre parênteses, separadas por vírgula.

```python
class ClasseFilha(ClasseMae1, ClasseMae2, ClasseMae3):
    pass
```

#### **Exemplo Prático: O Pato-Robô**


Imagine que estamos construindo um sistema para controlar diferentes tipos de robôs. Temos uma classe para robôs que andam e outra para robôs que voam.

- **Classe Base `Robo`**

````python
class Robo:
    """
    Superclasse genérica para todos os tipos de robôs.
    Define atributos e comportamentos comuns.
    """
    def __init__(self, nome: str):
        self.nome = nome
        print(f"Robô '{self.nome}' inicializado.")

    def identificar(self):
        print(f"Eu sou o robô {self.nome}.")
````

- **Subclasses Herdando de `Robo`**

````python
class RoboAndador(Robo):
    """
    Robô com capacidade de andar em superfícies.
    """
    def andar(self):
        print(f"{self.nome} está andando pelo chão.")


class RoboVoador(Robo):
    """
    Robô com capacidade de voar.
    """
    def voar(self):
        print(f"{self.nome} está voando pelos céus.")
````

- **`PatoRobo`**

````python
class PatoRobo(RoboAndador, RoboVoador):
    """
    Um robô que combina as habilidades de andar e voar.
    Herda de RoboAndador e RoboVoador.
    """
    def __init__(self, nome):
        # Para herança múltipla, é comum chamar o __init__ de cada pai explicitamente
        RoboAndador.__init__(self, nome)
        RoboVoador.__init__(self, nome)
        print("PatoRobo totalmente operacional!")

    def grasnar(self):
        print(f"{self.nome} faz: Quack! Quack! (versão metálica)")
````

Vamos testar nosso `PatoRobo`:

````python
# Criando um PatoRobo
pato_bot = PatoRobo("Duckinator 3000")
````

````python
# Ele pode usar métodos de ambas as classes mãe
pato_bot.andar()
pato_bot.voar()
````

````python
# E também tem seu próprio método
pato_bot.grasnar()
````

**A Ordem de Resolução de Métodos (MRO)**


Um desafio da **herança múltipla** ocorre quando diferentes classes mãe possuem métodos com o mesmo nome: qual deles a classe filha deve executar?
O Python resolve isso de forma determinística utilizando a **Method Resolution Order (MRO)**, ou **Ordem de Resolução de Métodos**. A MRO define a sequência em que as classes base são pesquisadas ao buscar um método. Em geral, a busca segue **da esquerda para a direita** na lista de classes mãe e **de baixo para cima** na hierarquia de herança.

Você pode inspecionar a MRO de qualquer classe usando o atributo especial `__mro__` ou o método `.mro()`.





````python
print(PatoRobo.__mro__)
````

Vamos interpretar passo a passo:

```python
(<class '__main__.PatoRobo'>,
 <class '__main__.RoboAndador'>,
 <class '__main__.RoboVoador'>,
 <class '__main__.Robo'>,
 <class 'object'>)
```

1. `PatoRobo` – a própria classe filha que estamos instanciando.
2. `RoboAndador` – a primeira classe mãe (esquerda para direita) na qual Python procura métodos não encontrados em PatoRobo.
3. `RoboVoador` – a segunda classe mãe.
4. `Robo` – a classe base comum a `RoboAndador` e `RoboVoador`.
5. `object` – a classe raiz de todas as classes em Python.
