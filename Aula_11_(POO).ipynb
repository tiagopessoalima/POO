# **Polimorfismo**

Capacidade de objetos de diferentes classes responderem à mesma mensagem de formas distintas. O polimorfismo permite escrever código mais genérico e flexível, onde o mesmo método pode se comportar de maneira diferente dependendo do objeto que o invoca.

**Objetivo da Semana:** Compreender o conceito de polimorfismo e suas diferentes formas de implementação em Python. Desenvolver a capacidade de utilizar sobrescrita de métodos para especializar comportamentos em hierarquias de classes, aplicar o princípio de duck typing para criar código mais flexível, e implementar métodos especiais para enriquecer a semântica dos objetos.

**Nota:** O polimorfismo permite tratar objetos de diferentes tipos de maneira uniforme, desde que respondam às mesmas mensagens. Em Python, isso é facilitado pelo duck typing - "se anda como um pato e grasna como um pato, então deve ser um pato".

## **Sobrescrita de Métodos: Especializando Comportamentos**

A sobrescrita de métodos (*method overriding*) ocorre quando uma subclasse fornece uma implementação específica para um método que já existe em sua superclasse. Isso permite que classes derivadas especializem o comportamento herdado.

### **Exemplo Prático: Sistema de Pagamentos**

````python
class FormaPagamento:
    """
    Classe base para todas as formas de pagamento.
    """
    def processar_pagamento(self, valor: float) -> bool:
        """
        Processa um pagamento genérico.

        Args:
            valor (float): Valor a ser pago

        Returns:
            bool: True se o pagamento foi processado com sucesso
        """
        print(f"Processando pagamento genérico de R$ {valor:.2f}")
        return True

    def estornar_pagamento(self, valor: float) -> bool:
        """
        Estorna um pagamento genérico.

        Args:
            valor (float): Valor a ser estornado

        Returns:
            bool: True se o estorno foi processado com sucesso
        """
        print(f"Estornando pagamento genérico de R$ {valor:.2f}")
        return True
````

Agora, vamos criar formas específicas de pagamento que sobrescrevem o comportamento padrão:

````python
class CartaoCredito(FormaPagamento):
    """
    Forma de pagamento com cartão de crédito.
    """
    def __init__(self, numero: str, titular: str, cvv: str):
        self.numero = numero
        self.titular = titular
        self.cvv = cvv

    def processar_pagamento(self, valor: float) -> bool:
        """
        Processa pagamento com cartão de crédito.
        Sobrescreve o método da classe base.
        """
        print(f"Processando pagamento de R$ {valor:.2f} no cartão {self.numero[-4:]}")
        print(f"Titular: {self.titular}")
        # Simulação de processamento
        return True

    def estornar_pagamento(self, valor: float) -> bool:
        """
        Estorna pagamento com cartão de crédito.
        Sobrescreve o método da classe base.
        """
        print(f"Estornando R$ {valor:.2f} para o cartão {self.numero[-4:]}")
        print("Estorno pode levar até 5 dias úteis")
        return True


class PayPal(FormaPagamento):
    """
    Forma de pagamento com PayPal.
    """
    def __init__(self, email: str):
        self.email = email

    def processar_pagamento(self, valor: float) -> bool:
        """
        Processa pagamento com PayPal.
        Sobrescreve o método da classe base.
        """
        print(f"Processando pagamento de R$ {valor:.2f} via PayPal")
        print(f"Email: {self.email}")
        print("Redirecionando para página do PayPal...")
        return True

    def estornar_pagamento(self, valor: float) -> bool:
        """
        Estorna pagamento com PayPal.
        Sobrescreve o método da classe base.
        """
        print(f"Estornando R$ {valor:.2f} para conta PayPal {self.email}")
        print("Estorno processado instantaneamente")
        return True


class BoletoBancario(FormaPagamento):
    """
    Forma de pagamento com boleto bancário.
    """
    def processar_pagamento(self, valor: float) -> bool:
        """
        Processa pagamento com boleto bancário.
        Sobrescreve o método da classe base.
        """
        print(f"Gerando boleto de R$ {valor:.2f}")
        print("Código de barras: 34191.79001 01043.510047 91020.150008 8 84410000015000")
        print("Vencimento: 30/11/2024")
        return True

    def estornar_pagamento(self, valor: float) -> bool:
        """
        Boleto não permite estorno direto.
        Sobrescreve o método da classe base.
        """
        print("Boleto bancário não permite estorno automático")
        print("Entre em contato com o suporte para reembolso")
        return False
````

Testando a Sobrescrita de Métodos

````python
# Criando diferentes formas de pagamento
cartao = CartaoCredito("1234 5678 9012 3456", "João Silva", "123")
paypal = PayPal("joao.silva@email.com")
boleto = BoletoBancario()

# Lista heterogênea de formas de pagamento
formas_pagamento = [cartao, paypal, boleto]

print("=== PROCESSANDO PAGAMENTOS ===")
valor = 150.50

for forma in formas_pagamento:
    print(f"\n--- {forma.__class__.__name__} ---")
    forma.processar_pagamento(valor)

print("\n=== ESTORNANDO PAGAMENTOS ===")
for forma in formas_pagamento:
    print(f"\n--- {forma.__class__.__name__} ---")
    forma.estornar_pagamento(valor)
````

## **Duck Typing: A Filosofia Pythonica**

Em Python, não precisamos que objetos herdem de uma classe base comum para serem usados polimorficamente. Basta que implementem os métodos necessários - isso é conhecido como *duck typing*.


### **Exemplo: Sistema de Notificações**

Vamos desenvolver diferentes tipos de notificadores. Eles não terão uma hierarquia comum, mas ainda assim poderão ser usados de forma polimórfica, como se pertencessem ao mesmo tipo.


````python
class EmailNotifier:
    """
    Envia notificações por email.
    """
    def __init__(self, email: str):
        self.email = email

    def enviar(self, mensagem: str) -> bool:
        """
        Envia notificação por email.
        """
        print(f"📧 Enviando email para {self.email}")
        print(f"   Assunto: Notificação Importante")
        print(f"   Mensagem: {mensagem}")
        return True
````

````python
class SMSNotifier:
    """
    Envia notificações por SMS.
    """
    def __init__(self, telefone: str):
        self.telefone = telefone

    def enviar(self, mensagem: str) -> bool:
        """
        Envia notificação por SMS.
        """
        print(f"📱 Enviando SMS para {self.telefone}")
        print(f"   Mensagem: {mensagem[:50]}...")  # Limita tamanho do SMS
        return True
````

````python
class PushNotifier:
    """
    Envia notificações push.
    """
    def __init__(self, dispositivo_id: str):
        self.dispositivo_id = dispositivo_id

    def enviar(self, mensagem: str) -> bool:
        """
        Envia notificação push.
        """
        print(f"🔔 Enviando push para dispositivo {self.dispositivo_id}")
        print(f"   Título: Nova Notificação")
        print(f"   Corpo: {mensagem}")
        return True
````

````python
# Classe que NÃO segue o padrão esperado
class FaxNotifier:
    """
    Envia notificações por fax (não implementa 'enviar').
    """
    def __init__(self, numero_fax: str):
        self.numero_fax = numero_fax

    def transmitir(self, mensagem: str) -> bool:
        """
        Transmite fax - método com nome diferente.
        """
        print(f"📠 Enviando fax para {self.numero_fax}")
        print(f"   Documento: {mensagem}")
        return True
````

Testando *Duck Typing*

````python
def enviar_notificacao_para_todos(notificadores: list, mensagem: str):
    """
    Função que demonstra duck typing.
    Funciona com qualquer objeto que tenha método 'enviar'.
    """
    print(f"\n=== ENVIANDO NOTIFICAÇÃO: '{mensagem}' ===")

    for notificador in notificadores:
        try:
            # Duck typing em ação: não importa o tipo, só importa se tem 'enviar'
            notificador.enviar(mensagem)
        except AttributeError:
            print(f"❌ {notificador.__class__.__name__} não possui método 'enviar'")

# Criando notificadores
email = EmailNotifier("cliente@email.com")
sms = SMSNotifier("(11) 99999-9999")
push = PushNotifier("device_abc123")
fax = FaxNotifier("(11) 3333-4444")

# Lista heterogênea - duck typing em ação!
notificadores = [email, sms, push, fax]

# Testando
enviar_notificacao_para_todos(notificadores, "Sua fatura vence em 5 dias!")
````

> **Princípio do Duck Typing:** "Se parece com um pato e grasna como um pato, então deve ser um pato." Em Python, se um objeto tem o método que estamos chamando, ele pode ser usado no contexto polimórfico

### **Métodos Especiais (Métodos Mágicos)**


Os métodos especiais (também chamados de métodos mágicos ou *dunder methods*) permitem que nossas classes se integrem melhor com a linguagem Python, suportando operações como comparações, representações em string, entre outras.

### **Por que usar Métodos Especiais?**


- **Integração Nativa:** Permitem que seus objetos funcionem com funções e operadores *built-in* do Python

- **Código Mais Legível:** `obj1` + `obj2` é mais claro que `obj1.somar(obj2)`

- **Consistência:** Seus objetos se comportam como os tipos nativos do Python

- **Controle Flexível:** Você define como suas classes respondem a operações comuns

### **Principais Métodos Especiais**

#### `__init__` vs `__new__`


Já conhecemos `__init__` (inicializador), mas existe também `__new__` (construtor):

````python
class MinhaClasse:
    def __new__(cls, *args, **kwargs):
        """Cria uma nova instância - chamado ANTES do __init__"""
        print("Criando nova instância...")
        instance = super().__new__(cls)
        return instance

    def __init__(self, valor):
        """Inicializa a instância - chamado DEPOIS do __new__"""
        print("Inicializando instância...")
        self.valor = valor

obj = MinhaClasse(10)
# Saída:
# Criando nova instância...
# Inicializando instância...
````

- **cls — referência à classe, não à instância:** Em métodos de classe (como __new__), o primeiro parâmetro é cls — ele representa a própria classe que está sendo instanciada. É análogo ao self nos métodos de instância, mas no nível da classe.

- **\*args — argumentos posicionais variáveis:** Empacota todos os argumentos posicionais recebidos. É uma tupla com os valores passados ao criar o objeto. Exemplo:

  ```
  obj = MinhaClasse(10, 20, 30)
  Dentro de __new__, args será (10, 20, 30)
  ```

- **\*\*kwargs — argumentos nomeados variáveis:** Empacota todos os argumentos nomeados (*keyword arguments*). É um dicionário, onde as chaves são os nomes dos parâmetros. Exemplo:
  ```
  obj = MinhaClasse(valor=10, ativo=True)
  Dentro de __new__, kwargs será {'valor': 10, 'ativo': True}
  ```


> Nota: Normalmente, você não precisa sobrescrever `__new__`. Ele é usado em casos especiais, como:
> 1. Implementar singletons (garantir que só exista uma instância da classe);
> 2. Controlar ou modificar a criação de instâncias, como em classes imutáveis (int, str, tuple);
> 3. Customizar metaclasses ou frameworks que gerenciam instâncias automaticamente.

Exemplo com tipo imutável (tuple)

````python
class MinhaTupla(tuple):
    def __new__(cls, valores):
        print("Criando a tupla")
        return super().__new__(cls, valores)

    def __init__(self, valores):
        print("Inicializando (não altera a tupla)")
````

#### **Representação em String: `__str__` e `__repr__`**


Controlam como um objeto é representado em forma de texto — tanto para humanos quanto para desenvolvedores.

````python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def __str__(self):
        """Representação legível para humanos"""
        return f"Pessoa: {self.nome}, {self.idade} anos"

    def __repr__(self):
        """Representação precisa para desenvolvedores"""
        return f"Pessoa('{self.nome}', {self.idade})"

p = Pessoa("Ana", 25)
````

Quando usamos `print(p)`: O Python chama `__str__`.

#### **Métodos para Comparação**

Vamos criar uma classe Fraction (fração) para demonstrar vários métodos especiais:

````python
class Fraction:
    def __init__(self, numerador, denominador=1):
        self.n = numerador
        self.d = denominador

    def __str__(self):
        return f"{self.n}/{self.d}"

    def __repr__(self):
        return f"Fraction({self.n}, {self.d})"

    # Métodos de comparação
    def __eq__(self, other): # Equal (igual)
        """Verifica igualdade (==)"""
        if isinstance(other, Fraction):
            return self.n * other.d == self.d * other.n
        return False

    def __lt__(self, other): # Less Than (menor que)
        """Verifica se é menor (<)"""
        if isinstance(other, Fraction):
            return self.n * other.d < self.d * other.n
        return NotImplemented

    def __le__(self, other): # Less or Equal (menor ou igual)
        """Verifica se é menor ou igual (<=)"""
        return self < other or self == other

    def __gt__(self, other): # Greater Than (maior que)
        """Verifica se é maior (>)"""
        return not (self <= other)

    def __ge__(self, other): # Greater or Equal (maior ou igual)
        """Verifica se é maior ou igual (>=)"""
        return not (self < other)

# Testando
f1 = Fraction(1, 2)
f2 = Fraction(2, 4)
f3 = Fraction(3, 4)

print(f1 == f2)  # True - 1/2 == 2/4
print(f1 < f3)   # True - 1/2 < 3/4
print(f3 > f2)   # True - 3/4 > 1/2
````

#### **Métodos Aritméticos**

Continuando com nossa classe `Fraction`

````python
class ArithmeticFraction(Fraction):

    # Métodos aritméticos
    def __add__(self, other): # Add (adicionar, somar)
        """Soma duas frações (+)"""
        if isinstance(other, Fraction):
            novo_n = self.n * other.d + other.n * self.d
            novo_d = self.d * other.d
            return Fraction(novo_n, novo_d)
        return NotImplemented

    def __sub__(self, other): # Sub (subtrair)
        """Subtrai duas frações (-)"""
        if isinstance(other, Fraction):
            novo_n = self.n * other.d - other.n * self.d
            novo_d = self.d * other.d
            return Fraction(novo_n, novo_d)
        return NotImplemented

    def __mul__(self, other): # Mul (multiplicar)
        """Multiplica duas frações (*)"""
        if isinstance(other, Fraction):
            return Fraction(self.n * other.n, self.d * other.d)
        return NotImplemented

    def __truediv__(self, other): # True Div (divisão “real”, não truncada)
        """Divide duas frações (/)"""
        if isinstance(other, Fraction):
            return Fraction(self.n * other.d, self.d * other.n)
        return NotImplemented

# Testando operações
f1 = ArithmeticFraction(1, 2)
f2 = ArithmeticFraction(1, 3)

soma = f1 + f2          # 1/2 + 1/3 = 5/6
subtracao = f1 - f2     # 1/2 - 1/3 = 1/6
multiplicacao = f1 * f2 # 1/2 * 1/3 = 1/6
divisao = f1 / f2       # 1/2 ÷ 1/3 = 3/2

print(f"{f1} + {f2} = {soma}")
print(f"{f1} - {f2} = {subtracao}")
print(f"{f1} * {f2} = {multiplicacao}")
print(f"{f1} / {f2} = {divisao}")
````

#### **Métodos para Coleções**


Vamos criar uma classe que se comporta como uma coleção:

````python
class Playlist:
    def __init__(self, nome, musicas):
        self.nome = nome
        self.musicas = list(musicas)

    def __len__(self):
        """Retorna o número de músicas na playlist"""
        return len(self.musicas)

    def __getitem__(self, index):
        """Permite indexação: playlist[0]"""
        return self.musicas[index]

    def __contains__(self, musica):
        """Permite verificação com 'in'"""
        return musica in self.musicas

    def __iter__(self):
        """Permite iteração com 'for'"""
        return iter(self.musicas)

# Testando
minha_playlist = Playlist("Rock Classics", ["Bohemian Rhapsody", "Stairway to Heaven", "Hotel California"])

print(len(minha_playlist))              # 3
print(minha_playlist[1])                # Stairway to Heaven
print("Hotel California" in minha_playlist)  # True

for musica in minha_playlist:
    print(f"🎵 {musica}")
````

#### **Context Managers: `__enter__` e `__exit__`**


Métodos especiais para gerenciamento de recursos:

````python
class GerenciadorArquivo:
    def __init__(self, nome_arquivo, modo):
        self.nome_arquivo = nome_arquivo
        self.modo = modo
        self.arquivo = None

    def __enter__(self):
        """Abre o recurso"""
        print(f"Abrindo arquivo {self.nome_arquivo}")
        self.arquivo = open(self.nome_arquivo, self.modo)
        return self.arquivo

    def __exit__(self, exc_type, exc_val, exc_tb):
        """Fecha o recurso, mesmo com exceções"""
        print(f"Fechando arquivo {self.nome_arquivo}")
        if self.arquivo:
            self.arquivo.close()
        # Retorna False para propagar exceções, True para suprimir
        return False

# Usando como context manager
with GerenciadorArquivo("exemplo.txt", "w") as arquivo:
    arquivo.write("Olá, métodos especiais!")
# O arquivo é fechado automaticamente aqui
````

### **Comparação Python vs Java**


**Python (Métodos Especiais)**

````python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def __str__(self):
        return f"{self.nome} ({self.idade} anos)"

    def __eq__(self, other):
        return self.nome == other.nome and self.idade == other.idade

p1 = Pessoa("João", 30)
print(p1)  # Chama automaticamente __str__
````

**Java (Métodos Convencionais)**

```java
public class Pessoa {
    private String nome;
    private int idade;
    
    public Pessoa(String nome, int idade) {
        this.nome = nome;
        this.idade = idade;
    }
    
    @Override
    public String toString() {
        return nome + " (" + idade + " anos)";
    }
    
    @Override
    public boolean equals(Object obj) {
        if (this == obj) return true;
        if (obj == null || getClass() != obj.getClass()) return false;
        Pessoa other = (Pessoa) obj;
        return idade == other.idade && nome.equals(other.nome);
    }
}

Pessoa p1 = new Pessoa("João", 30);
System.out.println(p1);  // Chama automaticamente toString()
```

**Principais diferenças:**
- Python: Métodos com nomes padronizados (`__str__`, `__eq__`)
- Java: Anotações `@Override` e métodos com nomes convencionais (`toString`, `equals`)
- Python: Mais métodos especiais disponíveis
- Java: Mais verboso, mas mais explícito
